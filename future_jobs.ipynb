```python
#Salary Range Based on Industry
result = pd.read_sql("""SELECT 'Industry' AS category, 
                          industry AS group_name, 
                          MIN(salary_usd) AS min_salary, 
                          MAX(salary_usd) AS max_salary, 
                          MAX(salary_usd) - MIN(salary_usd) AS salary_range 
                    FROM jobs GROUP BY industry
                        UNION ALL
                      SELECT 'Job Title' AS category, 
                            job_title  AS group_name, 
                            MIN(salary_usd) AS min_salary, 
                            MAX(salary_usd) AS max_salary, 
                            MAX(salary_usd) - MIN(salary_usd) AS salary_range 
                    FROM jobs GROUP BY job_title
                    ORDER BY category, salary_range DESC;""",
                     con = cnn)
```

```python
result
```

```python
#Salary Gap between Remote and Non Remote Jobs
result = pd.read_sql("""SELECT remote_option,
                           COUNT(*) AS total_jobs,
                           ROUND(AVG(salary_usd), 2) AS avg_salary,
                           MIN(salary_usd) AS min_salary,
                           MAX(salary_usd) AS max_salary
                       FROM jobs
                       GROUP BY remote_option
                       ORDER BY avg_salary DESC """,
                      con = cnn)
```

```python
result
```

```python
#Salary Gap between Remote and Non Remote Jobs
result = pd.read_sql("""SELECT job_title,
                           ROUND(AVG(CASE WHEN remote_option = 'Yes' THEN salary_usd END), 2) AS avg_remote_salary,
                           ROUND(AVG(CASE WHEN remote_option = 'No'  THEN salary_usd END), 2) AS avg_onsite_salary,
                           ROUND(AVG(CASE WHEN remote_option = 'Yes' THEN salary_usd END) -
                           AVG(CASE WHEN remote_option = 'No'  THEN salary_usd END), 2) AS salary_gap
                        FROM jobs
                        GROUP BY job_title
                        ORDER BY salary_gap DESC""",
                        con = cnn)
```

```python
result
```

```python
#City with Specific Industries
result = pd.read_sql("""SELECT 
                        location,
                        industry,
                        COUNT(*) AS total_jobs
                        FROM jobs
                        GROUP BY location, industry
                        ORDER BY industry, total_jobs DESC""",
                        con = cnn)
```

```python
#Top 10 Demanding Skills
result = pd.read_sql("""SELECT skill,
                        COUNT(*) AS demand_count,
                        ROUND(AVG(salary_usd), 2) AS avg_salary,
                        MIN(salary_usd) AS min_salary,
                        MAX(salary_usd) AS max_salary
                        FROM (
                                SELECT salary_usd,
                                TRIM(value) AS skill
                                FROM jobs, json_each('["' || REPLACE(skills_required, ', ', '","') || '"]')
                                )
                        GROUP BY skill
                        ORDER BY avg_salary DESC
                        LIMIT 10""",
                        con = cnn)
```
